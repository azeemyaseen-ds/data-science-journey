#### 1. You load a Kaggle CSV and a column called `age` prints as `'25'`, `'31'`, `'NaN'` — all strings. Write the logic to convert valid entries to `int` and explain what would break `int()` on `'NaN'`

```python
import pandas as pd
data = {
    "user_id":range(1,6),
    "name":["John", "Charlie", "Bob", "David", "Eva"],
    "age":["25", "32", "NaN", "30", "NaN"] 
}

df = pd.DataFrame(data)

print(df["age"].dtype)

# df["age"] = df["age"].astype("int") => The code breaks & through a ValueError because it is not possible for python to convert
# text into numeric value
```

### Corrections - 2026-09-16

```python
import pandas as pd
data = {
    "user_id":range(1,6),
    "name":["John", "Charlie", "Bob", "David", "Eva"],
    "age":["25", "32", "NaN", "30", "NaN"] 
}

df = pd.DataFrame(data)

df["age"] = pd.to_numeric(df["age"], errors="coerce")
print(df["age"])
print(df["age"].dtype)

# errors="coerce" turns unparseable entries ('NaN' string) into real NaN (float),
# instead of crashing the whole column like astype("int") does
```

#### 2. You have `price = "1,299.50"` (a price with a comma, straight from a scraped dataset). Walk through, step by step, what you'd need to strip/convert before `float()` would accept it — don't write the full solution yet, just list the transformation steps in order

```python
# a raw value
price = "1,299.50"

# 1. remove any trailing spaces
price = price.strip()

# 2. replace the comma with the empty string otherwise the float() can through ValueError
price = price.replace(",", "")

# 3. Pass the cleaned numeric string to convert it into float()
print(float(price))
```

```python
import numpy as np

scores = [88, 91.5, 76, 100.0]

print(type(scores[0]))
print(type(scores[1]))

scores = np.array(scores)
print(scores)

print(type(scores[0]))
print(type(scores[1]))

# python list can store multiple datatypes but numpy array can have a uniform datatype for all
# the values because numpy array are designed for high performance mathematical computation
# where we can perform vectorized operations on scale
```
