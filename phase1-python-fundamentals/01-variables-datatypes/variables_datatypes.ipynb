#### 2. You have `price = "1,299.50"` (a price with a comma, straight from a scraped dataset). Walk through, step by step, what you'd need to strip/convert before `float()` would accept it — don't write the full solution yet, just list the transformation steps in order

```python
# a raw value
price = "1,299.50"

# 1. remove any trailing spaces
price = price.strip()

# 2. replace the comma with the empty string otherwise the float() can through ValueError
price = price.replace(",", "")

# 3. Pass the cleaned numeric string to convert it into float()
print(float(price))
```

#### 1. A Kaggle dataset has a column `is_active` stored as the strings `"1"` and `"0"` (not `True`/`False`). Walk through what type it currently is, what type you actually want it as for a model, and what conversion path gets you there — you can't go straight `str → bool` safely, so what's the intermediate step and why?

```python
import pandas as pd

data ={
    "user_id": range(1,6),
    "name": ["John", "Charlie", "Bob", "David", "Eva"],
    "is_active": ["0", "0", "1", "0", "1"]
}

df = pd.DataFrame(data)

# what type it currently is?
print(df["is_active"].dtype)

# what type you actually want it as for a model?
# i want numerical binary data type because it help me to figure out the impact of the feature on the target variable with groupby
# method such as df.groupby("feature")["is_active"].mean()

# what conversion path gets you there?
df["is_active"] = df["is_active"].astype("int")
print(df["is_active"].dtype)
```

What data type does Python infer for each individually, and if you dump the whole column into one Python list, what type ends up governing the list for arithmetic purposes? Explain, don't just name it

```python
import pandas as pd

temperatures = pd.Series([101, 75, 98.6, 84, 105])

print(temperatures)

# what type ends up governing the list for arithmetic purposes
# when the column contain mixed datatypes (int or float) then python knows that every column must have uniform datatype,
# So python implicitly convert all the values to float because the float can represent the whole number but the whole number
# don't represent the float number
```
